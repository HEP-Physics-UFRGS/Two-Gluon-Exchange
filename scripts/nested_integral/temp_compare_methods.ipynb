```python
def add_iterative_curve(fig, x_data, y_data, title:str, x_axis_name:str, y_axis_name:str, curve_name:str):

    fig.add_trace(go.Scatter(
    x = x_data,
    y = y_data,
    mode='lines+markers', 
    name=curve_name,
    marker=dict(size=4))
)

    fig.update_layout(
        title=title,
        xaxis=dict(
            title = x_axis_name,
            type='log',
        ),
        yaxis=dict(
            title= y_axis_name,
            # range=[80, 120]
        ),
        showlegend=True,
        plot_bgcolor='white',
        hovermode='x unified'
    )
        
    fig.update_xaxes(gridcolor='lightgray')
    fig.update_yaxes(gridcolor='lightgray')

```

```python
import numpy as np
from scipy.integrate import quad, fixed_quad, nquad, romb
from scipy.special import j0
import mpmath as mp

# -------------------------------
# Parâmetros globais de integração
# -------------------------------
q_upper = 0.1
b_upper = 40

# -------------------------------
# Função de integral interna (sobre q)
# -------------------------------
def inner_integral(
    b, s, amp, method="quad", 
    N=1024, 
    epsabs_quad=1e-20, epsrel_quad=1e-20, limit_quad=100_000,
    n_fixed_quad=500,
    epsabs_nquad=1e-20, epsrel_nquad=1e-20, limit_nquad=2000,
    dps_mp=500
):
    if method in ["tanh_sinh", "mp_quad"]:
        mp.dps = dps_mp
        f = lambda q: q * mp.besselj(0, b*q)
        result = mp.quad(f, [0, q_upper], method='tanh-sinh')
        return float(result) * (1/s) * amp

    integrand = lambda q: q * j0(b*q)

    if method == "quad":
        result, _ = quad(integrand, 0, q_upper, limit=limit_quad, epsabs=epsabs_quad, epsrel=epsrel_quad)
    elif method == "fixed_quad":
        result, _ = fixed_quad(integrand, 0, q_upper, n=n_fixed_quad)
    elif method == "nquad":
        result, _ = nquad(
            integrand, [[0, q_upper]], 
            opts={'epsabs': epsabs_nquad, 'epsrel': epsrel_nquad, 'limit': limit_nquad}
        )
    elif method == "simpson":
        q = np.linspace(0, q_upper, N+1)
        f = integrand(q)
        result = (q_upper / N) * (np.sum(f) - 0.5*(f[0] + f[-1]))
    elif method == "romb":
        k = int(np.log2(N))
        N = 2**k
        q = np.linspace(0, q_upper, N+1)
        f = integrand(q)
        result = romb(f, dx=q_upper/N)
    else:
        raise ValueError(f"Unknown method {method}")

    return result * (1/s) * amp

# -------------------------------
# Integral externa (sobre b)
# -------------------------------
def outer_integrand(b, s, amp, method, **kwargs):
    if method in ["mp_quad", "tanh_sinh"]:
        chi = inner_integral(b, s, amp, method=method, **kwargs)
        return 1j * s * b * (1 - mp.exp(1j * chi))
    else:
        chi = np.array([inner_integral(bi, s, amp, method=method, **kwargs) for bi in np.atleast_1d(b)])
        return 1j * s * b * (1 - np.exp(1j * chi))

def outer_real(b, s, amp, method, **kwargs):
    val = outer_integrand(b, s, amp, method, **kwargs)
    return float(mp.re(val)) if isinstance(val, mp.mpc) else np.real(val)

def outer_imag(b, s, amp, method, **kwargs):
    val = outer_integrand(b, s, amp, method, **kwargs)
    return float(mp.im(val)) if isinstance(val, mp.mpc) else np.imag(val)

# -------------------------------
# Integral dupla
# -------------------------------
def compute_double_integral(
    s, amp, method="quad", 
    N=256, 
    epsabs_quad=1e-10, epsrel_quad=1e-10, limit_quad=100_000,
    n_fixed_quad=200,
    epsabs_nquad=1e-20, epsrel_nquad=1e-20, limit_nquad=2000,
    dps_mp=50
):
    kwargs = dict(
        N=N,
        epsabs_quad=epsabs_quad, epsrel_quad=epsrel_quad, limit_quad=limit_quad,
        n_fixed_quad=n_fixed_quad,
        epsabs_nquad=epsabs_nquad, epsrel_nquad=epsrel_nquad, limit_nquad=limit_nquad,
        dps_mp=dps_mp
    )

    if method in ["mp_quad", "tanh_sinh"]:
        f = lambda b: outer_integrand(b, s, amp, method, **kwargs)
        val = mp.quad(f, [0, b_upper], method='tanh-sinh')
        return complex(val)

    elif method in ["quad", "simpson", "romb"]:
        real_part, _ = quad(outer_real, 0, b_upper, args=(s, amp, method), limit=500, epsabs=epsabs_quad, epsrel=epsrel_quad)
        imag_part, _ = quad(outer_imag, 0, b_upper, args=(s, amp, method), limit=500, epsabs=epsabs_quad, epsrel=epsrel_quad)
        return real_part + 1j * imag_part

    elif method == "fixed_quad":
        real_part, _ = fixed_quad(outer_real, 0, b_upper, args=(s, amp, method), n=n_fixed_quad)
        imag_part, _ = fixed_quad(outer_imag, 0, b_upper, args=(s, amp, method), n=n_fixed_quad)
        return real_part + 1j * imag_part

    elif method == "nquad":
        real_part, _ = nquad(outer_real, [[0, b_upper]], args=(s, amp, method))
        imag_part, _ = nquad(outer_imag, [[0, b_upper]], args=(s, amp, method))
        return real_part + 1j * imag_part

    else:
        raise ValueError(f"Unknown method {method}")


# -------------------------------
# Loop de exemplo para comparar métodos
# -------------------------------
lst_amp_eik_quad = []
lst_amp_eik_fixed_quad = []
lst_amp_eik_nquad = []
lst_amp_eik_simpson = []
lst_amp_eik_romb = []
lst_amp_eik_tanh_sinh = []
lst_amp_eik_mp_quad = []

for s, amp in zip(lst_sqrt_s, lst_amp_born):
    s2 = s**2
    lst_amp_eik_quad.append(compute_double_integral(s2, amp, method="quad"))
    lst_amp_eik_fixed_quad.append(compute_double_integral(s2, amp, method="fixed_quad", N=200))
    lst_amp_eik_nquad.append(compute_double_integral(s2, amp, method="nquad"))
    lst_amp_eik_simpson.append(compute_double_integral(s2, amp, method="simpson", N=200))
    lst_amp_eik_romb.append(compute_double_integral(s2, amp, method="romb", N=256))
    lst_amp_eik_tanh_sinh.append(compute_double_integral(s2, amp, method="tanh_sinh"))
    lst_amp_eik_mp_quad.append(compute_double_integral(s2, amp, method="mp_quad"))
```
