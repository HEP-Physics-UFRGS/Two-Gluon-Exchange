```python
# Function to integrate
def f(x, y):
    return x**2 + y**2

# Inner integral over y from 0 to 2
def inner_integral(x):
    result, _ = fixed_quad(lambda y: f(x, y), 0, 2, n=5)
    return result

# Outer integral over x from 0 to 1
result, _ = fixed_quad(inner_integral, 0, 1, n=5)

print(f"Result of 2D integral: {result:.6f}")

```

```python
import numpy as np
from scipy.integrate import fixed_quad

# Função a ser integrada
def f(x, y):
    return np.sin(x * y) + x**2 * y

# Integral interna: de y = 0 até y = x
def inner_integral(x):
    # x = np.atleast_1d(x)  # Garante que x seja array
    result = np.array([fixed_quad(lambda y: f(xi, y), 0, xi, n=10)[0] for xi in x])
    return result   # Retorna escalar se x era escalar

# Integral externa: de x = 0 até x = pi
result, _ = fixed_quad(inner_integral, 0, np.pi, n=10)

print(f"Resultado da integral 2D: {result:.6f}")
```

```python
import numpy as np
from scipy.integrate import fixed_quad

# Parte dentro do expoente: integral de g(x, y) em y
def inner_integral(x):
    # g(x, y) = x*y + y^2
    return fixed_quad(lambda y: x * y + y**2, 0, 1, n=10)[0]

# Integral externa: exp(-inner_integral(x)) de x = 0 a 1
def outer_integrand(x):
    x = np.atleast_1d(x)
    result = np.array([np.exp(-inner_integral(xi)) for xi in x])
    return result if result.size > 1 else result[0]

# Integrar de 0 a 1
result, _ = fixed_quad(outer_integrand, 0, 1, n=10)

print(f"Resultado da integral 2D com expoente integral: {result:.6f}")
```

```python
import numpy as np
from scipy.integrate import fixed_quad

# Parte dentro do expoente: integral em y, para um dado x escalar
def inner_integral(x):
    return fixed_quad(lambda y: x * y + y**2, 0, 1, n=50)[0]

# Função externa — trata x como array
def outer_integrand(x_array):
    if np.isscalar(x_array):
        x_array = np.array([x_array])
    results = []
    for x in x_array:
        exp_term = np.exp(-inner_integral(x))
        multiplier = np.sin(np.pi * x) / (1 + x**2)
        results.append(multiplier * exp_term)
    return np.array(results)

# Integral externa de 0 a 1
result, _ = fixed_quad(outer_integrand, 0, 1, n=50)

print(f"Resultado da integral 2D com parte externa complexa: {result:.8f}")
```

```python
import numpy as np
from scipy.integrate import fixed_quad

# Integral 1: interna no expoente — depende de b e q
def inner_exponent_integral(b):
    integrand = lambda q: b * q + q**2 + np.sin(b * q)
    return fixed_quad(integrand, 0, 1, n=50)[0]

# Integral 2: outra dependência de q fora do expoente
def inner_multiplier_integral(b):
    integrand = lambda q: np.cos(np.pi * q * b) / (1 + q**2)
    return fixed_quad(integrand, 0, 1, n=50)[0]

# Função externa — integração em b, depende de q através das duas integrais acima
def outer_integrand(b_array):
    if np.isscalar(b_array):
        b_array = np.array([b_array])
    results = []
    for b in b_array:
        exp_term = np.exp(-inner_exponent_integral(b))  # integral no expoente
        q_term = inner_multiplier_integral(b)           # outra integral em q fora do expoente
        results.append(q_term * exp_term)
    return np.array(results)

# Integral externa sobre b
result, _ = fixed_quad(outer_integrand, 0, 1, n=50)

print(f"Resultado final com dependência em b e q: {result:.8f}")
```

```python
import numpy as np
from scipy.integrate import fixed_quad

# Integral INTERNA em q (igual ao anterior)
def inner_integral(b):
    integrand = lambda q: b * q + q**2 + np.sin(b * q)
    return fixed_quad(integrand, 0, 1, n=50)[0]

# Integral EXTERNA em b (agora com dependência em q)
def outer_integrand(b, q):
    I_int = inner_integral(b)  # Calcula a integral interna
    return np.cos(np.pi * b * q) * np.exp(-I_int)  # Termo com q no coseno

# Integral externa para um q fixo (exemplo: q = 0.5)
q_value = 0.5
result, _ = fixed_quad(lambda b: outer_integrand(b, q_value), 0, 1, n=50)
print(f"Resultado para q = {q_value}: {result:.8f}")

# Integral FINAL (integrando a externa sobre q)
def full_integral(q):
    integral_over_b, _ = fixed_quad(lambda b: outer_integrand(b, q), 0, 1, n=50)
    return integral_over_b

final_result, _ = fixed_quad(full_integral, 0, 1, n=50)
print(f"Resultado final (integral em b e q): {final_result:.8f}")
```
