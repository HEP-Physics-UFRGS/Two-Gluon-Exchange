```python
def sigma_tot_eik(s, amp):
    return (4*np.pi)/s * amp.imag * 0.389379323
```

```python
lst_b = [x for x in np.linspace(0, 22, 129)]
lst_s = [val**2 for val in lst_sqrt_s]

# -------------------------------
# Parâmetros globais de integração
# -------------------------------
q_upper = 0.2

epsabs = 1e-8
epsrel = 1e-8
limit = 10000

# -------------------------------
# Integral interna (sobre q)
# -------------------------------
def inner_integral(b, s, amp, epsabs=epsabs, epsrel=epsrel, limit=limit):
    integrand = lambda q: q * j0(b*q)
    result, _ = quad(integrand, 0, q_upper, epsabs=epsabs, epsrel=epsrel, limit=limit)
    return result * (1/s) * amp

def outer_integrand(b, s, chi):
    return 1j * s * b * (1 - np.exp(1j * chi))



# -------------------------------
# Integral externa (sobre b)
# -------------------------------
sum_chi = 0

lst_amp_eik_im = []
lst_sigma_tot_eik = []

for (b_val, s_val, amp_born_val) in zip(lst_b, lst_s, lst_amp_born):
    chi = inner_integral(b_val, s_val, amp_born_val)
    # print(chi)

    sum_chi += chi

    amp_eik_val = outer_integrand(b_val, s_val, sum_chi)
    amp_eik_im_val = amp_eik_val.imag
    lst_amp_eik_im.append(amp_eik_im_val)

    print(f'{amp_eik_val:.2e}')

    sigma_tot_eik_val = sigma_tot_eik(s_val, amp_eik_val)
    lst_sigma_tot_eik.append(sigma_tot_eik_val)
    print(f'{sigma_tot_eik_val}')
    print(50*'-')


```

```python
def add_iterative_curve(fig, x_data, y_data, 
                        curve_name:str=None, color:str='blue', line_type:str='lines+markers'):

    fig.add_trace(go.Scatter(
    x = x_data,
    y = y_data,
    mode=line_type, 
    name=curve_name,
    line=dict(
        color=color,
        width=2),
    marker=dict(size=4))
)
    fig.update_xaxes(gridcolor='lightgray')
    fig.update_yaxes(gridcolor='lightgray')
```
