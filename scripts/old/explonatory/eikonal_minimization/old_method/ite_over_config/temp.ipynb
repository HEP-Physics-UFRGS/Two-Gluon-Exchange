```python
from iminuit import Minuit
from iminuit.cost import LeastSquares
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
x = np.linspace(0, 10, 20) 
y = 3 * x**2 + 15 + np.random.randn(len(x))
err = np.ones(len(x))
plt.errorbar(x, y, err, fmt='.')
```

```python
def line(x, m, c): 
    return m * x**2 + c

chi2 = LeastSquares(x, y, err, line)
minuit = Minuit(chi2, m=1, c=0)  # Provide initial values for m and c
minuit.migrad()
```
