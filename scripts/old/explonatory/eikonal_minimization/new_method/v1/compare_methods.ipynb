```python

lower = 0
upper = 50

def f(q):
#     # Matches: BesselJ[0, 500*q] * I * BesselJ[1, 500*q] + Exp[I*500*q]/(1 + q)
#     return j0(500*q) * 1j * j1(500*q) + np.exp(1j * 500*q) / (1 + q)

    return j0(500*q) * j0(500*q) * j1(500*q) + np.exp(1j * 500*q) / (1 + q**0.1)
```

```python
# print('diff = abs(analytic - result)\n')

# ===================QUADPY===================
scheme = qp.c1.gauss_legendre(10000)
# Perform integration
result_quadpy = scheme.integrate(f, [lower, upper])
print(f'quadpy result: {result_quadpy}')
print(f'quadpy real = {result_quadpy.real}, imag = {result_quadpy.imag}')
print(f'quadpy error = {result_quadpy}')

print(100*'=')

result, error = qp.quad(f, lower, upper, limit = 10000)
print(f"Result: {result}")
print(f"Error estimate: {error}")


print(100*'=')

def f_imag(q):
    return np.imag(f(q))
# # # ===================QUAD (SCIPY)===================
val_quad, err_quad = quad(f, lower, upper, limit = 10000)
print(f'scipy quad: val = {val_quad}, err = {err_quad}')

val_quad, err_quad = quad(f_imag, lower, upper, limit = 10000)
print(f'scipy quad: val = {val_quad}, err = {err_quad}')

print(100*'=')

# # ===================FIXED QUAD (SCIPY)===================
val_fixed = fixed_quad(f, lower, upper, n = 10000)
print(f'fixed_quad (n=5): val = {val_fixed[0]}, err = {val_fixed[1]}')
print(f'fixed_quad (n=5) real = {val_fixed[0].real}, imag = {val_fixed[0].imag}')


```
